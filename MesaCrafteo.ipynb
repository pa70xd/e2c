```python
import requests
from dotenv import load_dotenv
import  os

load_dotenv()


def obtener_direccion(latitud, longitud):

    url = "https://maps.googleapis.com/maps/api/geocode/json"
    API_KEY = os.getenv('MAPS_KEY')


    params = {
        "latlng": f"{latitud},{longitud}",
        "key": API_KEY,
        "language": "es",
    }

    response = requests.get(url, params=params)

    if response.status_code != 200:
        raise Exception(f"Error HTTP: {response.status_code}")

    data = response.json()

    if data["status"] != "OK":
        raise Exception(
            f"Error de Google Maps: {data['status']} - "
            f"{data.get('error_message', '')}"
        )

    resultado = data["results"][0]

    direccion = resultado.get("formatted_address")

    datos = {
        "direccion": direccion,
        "codigo_postal": None,
        "colonia": None,
        "ciudad": None,
        "estado": None,
        "pais": None,
    }

    for componente in resultado["address_components"]:
        tipos = componente["types"]
        nombre = componente["long_name"]

        if "postal_code" in tipos:
            datos["codigo_postal"] = nombre

        elif "sublocality" in tipos or "neighborhood" in tipos:
            datos["colonia"] = nombre

        elif "locality" in tipos:
            datos["ciudad"] = nombre

        elif "administrative_area_level_1" in tipos:
            datos["estado"] = nombre

        elif "country" in tipos:
            datos["pais"] = nombre

    return datos


if __name__ == "__main__":

    latitud = 18.879256	
    longitud = -97.129475

    datos = obtener_direccion(latitud, longitud)

    print("\nResultado:")
    print(f"Dirección:     {datos['direccion']}")
    print(f"Código postal: {datos['codigo_postal']}")
    print(f"Colonia:       {datos['colonia']}")
    print(f"Ciudad:        {datos['ciudad']}")
    print(f"Estado:        {datos['estado']}")
    print(f"País:          {datos['pais']}")
```
